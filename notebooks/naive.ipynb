```python
import jax
from flock.env import EnvConfig, RandomPolicy, run_episode
from flock.env.rules import PredatorPrey
from flock.env.viz import animate, plot_stats
from flock.train.naive import ChasePrey, FleePredators
```

```python
env_cfg = EnvConfig(
    arena_size=10.0,
    dt=0.05,
    max_steps=500,
)

rules = PredatorPrey(
    n_predators=5,
    n_prey=20,
    max_speed_predator=1.0,
    max_accel_predator=8.0,
    max_speed_prey=1.5,
    max_accel_prey=1.5 * 8.0,
    catch_radius=0.3,
    k_teammates=5,
    k_opponents=5,
)
```

### Naive vs Random baseline

```python
# Baseline: both teams random
random_policies = tuple(RandomPolicy(tc) for tc in rules.teams)
sim_random = run_episode(env_cfg, rules, jax.random.key(0), random_policies)
```

```python
# Naive: chase/flee heuristics
naive_policies = (ChasePrey(rules.teams[0]), FleePredators(rules.teams[1]))
sim_naive = run_episode(env_cfg, rules, jax.random.key(0), naive_policies)
```

### Compare: random baseline

```python
_ = animate(sim_random)
```

```python
_ = plot_stats(sim_random)
```

### Naive chase/flee

```python
_ = animate(sim_naive)
```

```python
_ = plot_stats(sim_naive)
```
