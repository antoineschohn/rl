```python
import jax
from flock.env import EnvConfig, RandomPolicy, run_episode
from flock.env.rules import PredatorPrey
from flock.env.viz import animate, plot_stats
```

```python
env_cfg = EnvConfig(
    arena_size=10.0,
    dt=0.05,
    max_steps=500,
)

rules = PredatorPrey(
    n_predators=5,
    n_prey=20,
    max_speed_predator=1.0,
    max_accel_predator=8.0,
    max_speed_prey=1.5,
    max_accel_prey=1.5*8.0,
    catch_radius=0.3,
    k_teammates=5,
    k_opponents=5,
)
```

```python
policies = tuple(RandomPolicy(tc) for tc in rules.teams)
sim = run_episode(env_cfg, rules, jax.random.key(0), policies)
```

```python
_ = animate(sim)
```

```python
_ = plot_stats(sim)
```
