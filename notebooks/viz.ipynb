```python
import jax
from flock.env import EnvConfig, run_episode, random_policy
from flock.env.viz import animate
```

```python
config = EnvConfig(n_predators=5, n_prey=20, arena_size=10.0, max_steps=300)
sim = run_episode(
    config, jax.random.key(0),
    pred_policy=random_policy(config.n_predators),
    prey_policy=random_policy(config.n_prey),
)
animate(sim)
```
