```python
import jax
from flock.env import EnvConfig, RandomPolicy, run_episode
from flock.env.viz import animate, plot_stats
```

```python
config = EnvConfig(n_predators=5, n_prey=20, arena_size=10.0, max_steps=300)
sim = run_episode(
    config, jax.random.key(0),
    pred_policy=RandomPolicy(config.n_predators),
    prey_policy=RandomPolicy(config.n_prey),
)
```

```python
_ = animate(sim)
```

```python
_ = plot_stats(sim)
```
