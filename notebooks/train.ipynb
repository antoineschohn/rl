```python
import jax
from flock.env import EnvConfig, run_episode
from flock.env.rules import PredatorPrey
from flock.train import TrainConfig, train
from flock.env.viz import animate, plot_stats
```

```python
env_cfg = EnvConfig(max_steps=200)
rules = PredatorPrey(n_predators=5, n_prey=20, k_teammates=4, k_opponents=5)
train_cfg = TrainConfig(
    n_arenas=16,
    n_train_steps=300,
    n_epochs=4,
    n_minibatches=4,
    lr=3e-4,
)
```

```python
policies = train(env_cfg, rules, train_cfg, key=jax.random.key(0))
```

```python
sim = run_episode(env_cfg, rules, jax.random.key(99), policies)
```

```python
_ = animate(sim)
```

```python
_ = plot_stats(sim)
```
