```python
import jax
from flock.env import EnvConfig, run_episode
from flock.train import TrainConfig, train
from flock.env.viz import animate, plot_stats
```

```python
env_cfg = EnvConfig(
    n_predators=5, n_prey=20,
    arena_size=10.0,
    max_steps=100,
    k_teammates=4, k_opponents=5,
)
train_cfg = TrainConfig(
    n_arenas=16,
    n_train_steps=100,
    n_epochs=4,
    n_minibatches=4,
    lr=3e-4,
)
```

```python
pred_policy, prey_policy = train(env_cfg, train_cfg, key=jax.random.key(0))
```

```python
sim = run_episode(env_cfg, jax.random.key(99), pred_policy, prey_policy)
```

```python
_ = animate(sim)
```

```python
_ = plot_stats(sim)
```
